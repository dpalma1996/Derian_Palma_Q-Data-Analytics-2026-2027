#### <span style="color: rgb(230, 0, 126);"> Exercici 1 (2p)</span>

Crea una funció que rebi una llista de números com a paràmetre.
La funció ha de recórrer la llista i retornar una altra llista amb aquestes dades:
- Quantitat de números que hi ha a la llista
- Valor máxim de la llista

Després mostra per pantalla aquests dos valors.

DADES PER LA LLISTA:

40, 20, 33, 5, 14, 45, 87, 3, 10, 99

*Fes l'exercici sense utilitzar funcions de python com max(), min(), len(), count()*

```python
#escriu el teu codi python

# declaro desde el inicio la variable/lista
llista = [40, 20, 33, 5, 14, 45, 87, 3, 10, 99]


def func(list):
  # prefiero crear las variables que me servirán dentro de la función
  # las variables del contador y del máximo
  c = 0
  m = 0


  # aquí de momento uso len() para usarla con el rango,
  # NO la uso para sustituir el contador, solo para el bucle
  for i in range(len(list)):

    # este es mi contador
    c = c+1

    # esta lógica de comparación me servirá para encontrar el máximo
    # partimos de un máximo que es cero y comparamos cada vez
    # y si es mayor lo guardamos
    # no hace falta el caso contrario
    if list[i] > m:
      m = list[i]

  # devuelve ambos valores, el contador y el máximo
  return c, m

# llamamos a la función
# y almacenamos los valores devueltos por la función
# en dos variables de una vez
cant, max = func(llista)

# mostramos un mensaje con lo que nos solicitan
print(f"Quantitat de números que hi ha a la llista: {cant}. \nValor máxim de la llista: {max}.")
```

#### <span style="color: rgb(230, 0, 126);"> Exercici 4 (3,5p)</span>


En una empresa de logística, cada repartidor té assignades entregues amb una quantitat estimada de productes. Sovint, per diversos motius (errors, manca d'estoc, incidències...), la quantitat real entregada pot ser inferior a la prevista. Això afecta la qualitat del servei.

Objectiu:

Escriu una funció que, donat el dataset lliuraments.csv, faci el següent:

- Agrupar per repartidor i calcular els lliuraments de cadascú
- Agrupar per repartidor i calcular els lliuraments incomplets
- Calcular el % d'entregues incompletes (incompletes / totals * 100)
- Calcular la mitjana d'entregues incompletes de tots els repartidors
- Retornar el dataframe només amb els repartidors que estan per sobre de la mitjana

Crida la funció amb el dataframe lliuraments.csv i mostra els repartidors que estan per sobre de la mitjana.

----------------------------------------------------------------------------------------------------------------------

Ajuda: per "ajuntar" les series o dataframes que generis a les dues primeres agrupacions pots utilizar el mètode concat

nou_data_frame = pd.concat([serie1, serie2], axis=1)  
aquesta instrucció genera un dataframe "ajuntant" la serie1 i la serie2 (o si fosin els df1 o df2 també funcionaria).

Després, pots renombrar les columnes així:
nou_data_frame.columns = ['nom_columna1', 'nom_columna2']



```python
#escriu el teu codi python

import pandas as pd

data2 = pd.read_csv(
    "/content/sample_data/lliuraments-1.csv"
)

data2.head()


def func(df):

  df_nou = df.copy()


  # Agrupar per repartidor i calcular els lliuraments de cadascú
  totals = df_nou.groupby("repartidor").size()

  print("lliuraments de cadascú\n",totals)


  # Agrupar per repartidor i calcular els lliuraments incomplets
  df_incomplets = df_nou[
        df_nou["quantitat_entregada"] < df_nou["quantitat_prevista"]
    ]

  entregues_incomplets = df_incomplets.groupby("repartidor").size()

  print("\nlliuraments incomplets\n",entregues_incomplets)


  # Calcular el % d'entregues incompletes (incompletes / totals * 100)

  # primero habría que concatenar:

  resultat = pd.concat(
        [totals, entregues_incomplets],
        axis=1
    )

  # se nombran las columnas
  resultat.columns = [
      "totals",
      "entregues_incomplets"
  ]

  # los repartidores sin entregas incompletas aparecen con NaN.
  # se sustituyen esos valores por 0.
  resultat["entregues_incomplets"] = (
      resultat["entregues_incomplets"].fillna(0)
  )

  # por fin se hace el cálculo
  resultat["percentatge_incomplets"] = (
    resultat["entregues_incomplets"]
        / resultat["totals"]
        * 100
      )
  print("\n% d'entregues incompletes:\n",resultat["percentatge_incomplets"])


  # Calcular la mitjana d'entregues incompletes de tots els repartidors

  # se calcula la media del porcentaje de entregas incompletas
  mitjana = resultat["percentatge_incomplets"].mean()

  # se filtra los repartidores que están por encima de la media
  resultat_final = resultat[
      resultat["percentatge_incomplets"] > mitjana
  ]


  # Retornar el dataframe només amb els repartidors que estan per sobre de la mitjana
  return resultat_final

# se llama a la función
dada = func(data2)
print(dada)
```
